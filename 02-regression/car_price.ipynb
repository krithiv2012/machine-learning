```python
strings = df.select_dtypes('string').columns.to_list()
```

```python
for col in strings:
    df[col] = df[col].str.lower().str.replace(' ', '_')
```

```python
n = len(df)
n
```

```python
n_val = int(n * 0.2)
n_test = int(n* 0.2)
print(n_val)
print(n_test)
n_train = n - n_val - n_test
print(n_train)
```

```python
if n == n_train + n_val + n_test:
    print("Input data split successful without any missing records")
else:
    print("Missing records found while split the data")
```
