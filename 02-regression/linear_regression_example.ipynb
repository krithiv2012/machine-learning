```python
import numpy as np
xi = [453, 11, 86]
```

```python
w0 = 7.17
```

```python
w = [0.01,0.04, 0.002]
```

```python
def linear_regress(xi):
    n = len(xi)
    pred = w0
    for i in range(n):
        pred = pred + w[i] * xi[i]
    print(pred)
    return pred
```

```python
estimate= linear_regress(xi)
```

```python
estimated_price = np.expm1(estimate)
```
