Vector-Vector Multuplication

```python
u = np.array([2,4,5,6])
v = np.array([1,0,0,2])
```

```python
def vector_vector_multiplication(u, v):
    assert u.shape[0] == v.shape[0], "size not same"
    n = u.shape[0]
    result = 0.0
    for i in range(n):
        result = result + u[i] * v[i]
    #print(result)
    return result
        
```

```python
vector_vector_multiplication(u,v)
```

```python
def matrix_vector_multiplication(u,v):
    assert u.shape[1] == v.shape[0], "matrix row size and vector size not matching"
    n = u.shape[0]
    result = np.zeros(n)
    #print(result)
    for i in range(n):
        #print(u[i,:])
        #print(v)
        result[i] = vector_vector_multiplication(u[i], v)
    print(result)
```
